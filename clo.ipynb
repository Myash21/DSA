```python
# Fuzzy Sets
A = [0.2, 0.5, 0.7]
B = [0.3, 0.6, 0.4]
```

```python
# Union
union = [max(A[i], B[i]) for i in range(len(A))]
# Intersection
intersection = [min(A[i], B[i]) for i in range(len(A))]
# Complement
comp_A = [1 - x for x in A]
# Difference A - B
difference = [min(A[i], 1 - B[i]) for i in range(len(A))]
print("Union:", union)
print("Intersection:", intersection)
print("Complement of A:", comp_A)
print("Difference A-B:", difference)
```

```python
# Cartesian Product (Fuzzy Relation)
#R1 = [[min(a, b) for b in B] for a in A]
print("\nCartesian Product (Relation R1):")
R1 = []
for a in A:
    row = []
    for b in B:
        val = min(a, b)
        row.append(val)
    R1.append(row)
for row in R1:
    print(row)
```

```python
# Another Relation for Composition
C = [0.8, 0.4]
R2 = []
for b in B:
    row = []
    for c in C:
        val = min(b, c)
        row.append(val)
    R2.append(row)
print("\nRelation R2:")
for row in R2:
    print(row)
```

```python
composition = []
for i in range(len(R1)):
    row = []
    for j in range(len(R2[0])):
        values = []
        for k in range(len(R1)):
            values.append(min(R1[i][k], R2[k][j]))
        row.append(max(values))
    composition.append(row)

print("\nMax-Min Composition:")
for row in composition:
    print(row)
```
